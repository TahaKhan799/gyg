```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import mean_squared_error, r2_score, accuracy_score, classification_report

# =========================================================
# Assignment 1 & Mini Project 2: House Price Prediction (Linear Regression)
# =========================================================
print("--- 1. HOUSE PRICE PREDICTION (Linear Regression) ---")

# Load standard housing dataset
housing = fetch_california_housing(as_frame=True)
df_housing = housing.frame

# Define features and target variable
X_house = df_housing.drop(columns=['MedHouseVal'])
y_house = df_housing['MedHouseVal']

# Split dataset into training and testing sets (80% train, 20% test)
X_train_h, X_test_h, y_train_h, y_test_h = train_test_split(X_house, y_house, test_size=0.2, random_state=42)

# Train Linear Regression model
lin_reg = LinearRegression()
lin_reg.fit(X_train_h, y_train_h)

# Predict house prices
y_pred_h = lin_reg.predict(X_test_h)

# Evaluate performance (R^2 score and MSE)
r2 = r2_score(y_test_h, y_pred_h)
mse = mean_squared_error(y_test_h, y_pred_h)
print(f"R² Score: {r2:.4f}")
print(f"Mean Squared Error: {mse:.4f}")

# Plot predicted vs actual values
plt.figure(figsize=(7, 5))
plt.scatter(y_test_h, y_pred_h, alpha=0.3, color='royalblue')
plt.plot([y_test_h.min(), y_test_h.max()], [y_test_h.min(), y_test_h.max()], 'r--', lw=2)
plt.title('Mini Project 2: Predicted vs Actual House Prices')
plt.xlabel('Actual Prices')
plt.ylabel('Predicted Prices')
plt.grid(True, linestyle='--', alpha=0.6)
plt.savefig('house_price_prediction.png', dpi=300, bbox_inches='tight')
plt.show()

# =========================================================
# Assignment 2: Titanic Survival Prediction (Logistic Regression)
# =========================================================
print("\n--- 2. TITANIC SURVIVAL PREDICTION (Logistic Regression) ---")

# Load Titanic dataset from public source
titanic_url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df_titanic = pd.read_csv(titanic_url)

# Handle missing data and encode categorical feature ('Sex')
df_titanic['Age'] = df_titanic['Age'].fillna(df_titanic['Age'].median())
df_titanic['Sex_Encoded'] = df_titanic['Sex'].map({'male': 0, 'female': 1})

# Select feature columns and target
features = ['Pclass', 'Age', 'SibSp', 'Parch', 'Fare', 'Sex_Encoded']
X_tit = df_titanic[features].dropna()
y_tit = df_titanic.loc[X_tit.index, 'Survived']

# Split data
X_train_t, X_test_t, y_train_t, y_test_t = train_test_split(X_tit, y_tit, test_size=0.2, random_state=42)

# Train Logistic Regression model
log_reg = LogisticRegression(max_iter=1000)
log_reg.fit(X_train_t, y_train_t)

# Predict and evaluate
y_pred_t = log_reg.predict(X_test_t)
accuracy = accuracy_score(y_test_t, y_pred_t)
print(f"Logistic Regression Accuracy: {accuracy:.4f}")
print("\nClassification Report:")
print(classification_report(y_test_t, y_pred_t))
```
